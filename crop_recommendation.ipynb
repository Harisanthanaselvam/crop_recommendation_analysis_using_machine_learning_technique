```python
from sklearn.model_selection import train_test_split
```

```python
x= df.drop("Crop",axis=1)
y= df['Crop']
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
print(y_test.shape)
print(x_test.shape)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
import pandas as pd 

print("\nEvaluating RandomForestClassifier...\n")
rf = RandomForestClassifier(random_state=42)
rf.fit(x_train, y_train)
y_pred = rf.predict(x_test)

cm = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred, average='macro', zero_division=0)
rec = recall_score(y_test, y_pred, average='macro', zero_division=0)
f1 = f1_score(y_test, y_pred, average='macro', zero_division=0)

print("Confusion Matrix:\n", cm)
print(f"\nAccuracy Score: {acc:.2f}")
print(f"Precision: {prec:.2f}")
print(f"Recall: {rec:.2f}")
print(f"F1 Score: {f1:.2f}")
print("-" * 60)

```

```python

print("\nEvaluating SVC...\n")
svc = SVC()
svc.fit(x_train, y_train)
y_pred = svc.predict(x_test)

cm = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred, average='macro', zero_division=0)
rec = recall_score(y_test, y_pred, average='macro', zero_division=0)
f1 = f1_score(y_test, y_pred, average='macro', zero_division=0)

print("Confusion Matrix:\n", cm)
print(f"\nAccuracy Score: {acc:.2f}")
print(f"Precision: {prec:.2f}")
print(f"Recall: {rec:.2f}")
print(f"F1 Score: {f1:.2f}")
print("-" * 60)

```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


print(" Crop Encoding Mapping:")
print("0  -> Apple")
print("1  -> Banana")
print("2  -> Blackgram")
print("3  -> ChickPea")
print("4  -> Coconut")
print("5  -> Coffee")
print("6  -> Cotton")
print("7  -> Grapes")
print("8  -> Jute")
print("9  -> KidneyBeans")
print("10 -> Lentil")
print("11 -> Maize")
print("12 -> Mango")
print("13 -> MothBeans")
print("14 -> MungBean")
print("15 -> Muskmelon")
print("16 -> Orange")
print("17 -> Papaya")
print("18 -> PigeonPeas")
print("19 -> Pomegranate")
print("20 -> Rice")
print("21 -> Watermelon")



sns.set(style="whitegrid")

def get_crop_info_with_plots(crop_name):
    crop_data = df[df['Crop'].astype(str).str.lower() == crop_name.lower()]

    if crop_data.empty:
        print("Crop not found. Please check your spelling.")
        return

    print(f"\nGrowing Requirements for '{crop_name.capitalize()}':")
    print("-" * 60)

    features = ['Nitrogen', 'Phosphorus', 'Potassium', 'Temperature', 'Humidity', 'pH_Value', 'Rainfall']

    for feature in features:
        min_val = crop_data[feature].min()
        max_val = crop_data[feature].max()
        avg_val = crop_data[feature].mean()

    
        print(f" {feature}: {min_val:.2f} - {max_val:.2f} (Average: {avg_val:.2f})")

 
        plt.figure(figsize=(6, 4))
        sns.boxplot(x=crop_data[feature], color="skyblue")
        plt.title(f"{feature} Distribution for {crop_name.capitalize()}")
        plt.xlabel(feature)
        plt.grid(True)
        plt.tight_layout()
        plt.show()

    print("-" * 60)

user_crop = input("Enter the crop name to view requirements and plots: ")
get_crop_info_with_plots(user_crop)
```
